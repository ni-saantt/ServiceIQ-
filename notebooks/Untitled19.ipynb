```python
missing_summary = pd.DataFrame({
    "missing_count": (df == "?").sum(),
    "missing_percent": ((df == "?").sum() / len(df) * 100).round(2)
})

missing_summary = missing_summary[
    missing_summary["missing_count"] > 0
].sort_values("missing_percent", ascending=False)

display(missing_summary)
```

```python
drop_cols = [
    "caused_by",
    "vendor",
    "cmdb_ci",
    "rfc",
    "problem_id",
    "sys_created_by",
    "u_symptom"
]

df_clean = df.drop(columns=drop_cols)

print("Original columns:", len(df.columns))
print("Cleaned columns:", len(df_clean.columns))
print("\nRemaining columns:")
print(df_clean.columns.tolist())
```

```python
missing_clean = (df_clean == "?").sum().sort_values(ascending=False)

missing_clean[missing_clean > 0]
```

```python
fill_values = {
    "assigned_to": "Unassigned",
    "assignment_group": "Unassigned",
    "opened_by": "Unknown",
    "closed_code": "Unknown",
    "resolved_by": "Unknown",
    "subcategory": "Unknown",
    "category": "Unknown",
    "location": "Unknown",
    "caller_id": "Unknown"
}

df_clean = df_clean.replace("?", pd.NA)
df_clean = df_clean.fillna(fill_values)

print("Remaining missing values:")
print(df_clean.isna().sum()[df_clean.isna().sum() > 0])
```

```python
incident = df_clean[df_clean["number"] == "INC0019396"]

print("Number of records:", len(incident))

display(
    incident[
        [
            "number",
            "incident_state",
            "active",
            "reassignment_count",
            "reopen_count",
            "priority",
            "assignment_group",
            "opened_at",
            "sys_updated_at",
            "resolved_at",
            "closed_at"
        ]
    ].tail()
)
```

```python
df_clean = df_clean.sort_values(
    ["number", "sys_updated_at"]
).reset_index(drop=True)

print(df_clean[["number", "sys_updated_at"]].head())
```

```python
check_cols = [
    "priority",
    "impact",
    "urgency",
    "category",
    "subcategory",
    "assignment_group"
]

for col in check_cols:
    changes = df_clean.groupby("number")[col].nunique()
    print(f"{col}:")
    print("  Incidents with changes:", (changes > 1).sum())
    print("  Maximum different values in one incident:", changes.max())
```

```python
incident_df = df_clean.groupby("number").agg(
    opened_at=("opened_at", "first"),
    resolved_at=("resolved_at", "last"),
    closed_at=("closed_at", "last"),

    final_state=("incident_state", "last"),
    final_active=("active", "last"),

    reassignment_count=("reassignment_count", "last"),
    reopen_count=("reopen_count", "last"),
    sys_mod_count=("sys_mod_count", "last"),

    made_sla=("made_sla", "last"),

    priority=("priority", "last"),
    impact=("impact", "last"),
    urgency=("urgency", "last"),

    category=("category", "last"),
    subcategory=("subcategory", "last"),

    assignment_group=("assignment_group", "last"),
    assigned_to=("assigned_to", "last"),

    location=("location", "last"),
    contact_type=("contact_type", "first"),

    knowledge=("knowledge", "last"),
    u_priority_confirmation=("u_priority_confirmation", "last"),

    closed_code=("closed_code", "last"),
    resolved_by=("resolved_by", "last")
).reset_index()

print("Event records:", len(df_clean))
print("Unique incidents:", len(incident_df))

display(incident_df.head())
```

```python
incident_df["resolution_time_hours"] = (
    incident_df["resolved_at"] - incident_df["opened_at"]
).dt.total_seconds() / 3600

incident_df["closure_time_hours"] = (
    incident_df["closed_at"] - incident_df["opened_at"]
).dt.total_seconds() / 3600

display(
    incident_df[
        [
            "number",
            "opened_at",
            "resolved_at",
            "closed_at",
            "resolution_time_hours",
            "closure_time_hours"
        ]
    ].head()
)
```

```python
print("Negative resolution times:", (incident_df["resolution_time_hours"] < 0).sum())
print("Negative closure times:", (incident_df["closure_time_hours"] < 0).sum())

print("Missing resolution times:", incident_df["resolution_time_hours"].isna().sum())
print("Missing closure times:", incident_df["closure_time_hours"].isna().sum())
```

```python
display(
    incident_df[
        ["resolution_time_hours", "closure_time_hours"]
    ].describe()
)
```

```python
incident_df["is_reopened"] = incident_df["reopen_count"] > 0

incident_df["is_reassigned"] = incident_df["reassignment_count"] > 0

incident_df["sla_breached"] = incident_df["made_sla"] == False

incident_df["resolution_time_days"] = (
    incident_df["resolution_time_hours"] / 24
)

display(
    incident_df[
        [
            "number",
            "reopen_count",
            "is_reopened",
            "reassignment_count",
            "is_reassigned",
            "made_sla",
            "sla_breached",
            "resolution_time_days"
        ]
    ].head()
)
```

```python
sla_summary = incident_df["made_sla"].value_counts()

print("SLA Met:", sla_summary.get(True, 0))
print("SLA Not Met:", sla_summary.get(False, 0))

print(
    "SLA Compliance Rate:",
    round(incident_df["made_sla"].mean() * 100, 2),
    "%"
)

print(
    "SLA Breach Rate:",
    round(incident_df["sla_breached"].mean() * 100, 2),
    "%"
)
```

```python
team_sla = (
    incident_df
    .groupby("assignment_group")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum")
    )
    .reset_index()
)

team_sla["sla_breach_rate"] = (
    team_sla["sla_breaches"] /
    team_sla["total_incidents"] * 100
)

team_sla = team_sla.sort_values(
    "sla_breach_rate",
    ascending=False
)

display(team_sla.head(15))
```

```python
team_sla_50 = team_sla[
    team_sla["total_incidents"] >= 50
].sort_values(
    "sla_breach_rate",
    ascending=False
)

display(team_sla_50)
```

```python
priority_sla = (
    incident_df
    .groupby("priority")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum")
    )
    .reset_index()
)

priority_sla["sla_breach_rate"] = (
    priority_sla["sla_breaches"] /
    priority_sla["total_incidents"] * 100
)

display(
    priority_sla.sort_values("priority")
)
```

```python
priority_resolution = (
    incident_df
    .groupby("priority")
    .agg(
        total_incidents=("number", "count"),
        avg_resolution_hours=("resolution_time_hours", "mean"),
        median_resolution_hours=("resolution_time_hours", "median"),
        avg_closure_hours=("closure_time_hours", "mean")
    )
    .reset_index()
)

display(
    priority_resolution.sort_values("priority")
)
```

```python
category_sla = (
    incident_df
    .groupby("category")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum"),
        median_resolution_hours=("resolution_time_hours", "median")
    )
    .reset_index()
)

category_sla["sla_breach_rate"] = (
    category_sla["sla_breaches"] /
    category_sla["total_incidents"] * 100
)

category_sla = category_sla[
    category_sla["total_incidents"] >= 100
].sort_values(
    "sla_breach_rate",
    ascending=False
)

display(category_sla)
```

```python
reassignment_sla = (
    incident_df
    .groupby("is_reassigned")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum"),
        median_resolution_hours=("resolution_time_hours", "median")
    )
    .reset_index()
)

reassignment_sla["sla_breach_rate"] = (
    reassignment_sla["sla_breaches"] /
    reassignment_sla["total_incidents"] * 100
)

display(reassignment_sla)
```

```python
reopen_sla = (
    incident_df
    .groupby("is_reopened")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum"),
        median_resolution_hours=("resolution_time_hours", "median")
    )
    .reset_index()
)

reopen_sla["sla_breach_rate"] = (
    reopen_sla["sla_breaches"] /
    reopen_sla["total_incidents"] * 100
)

display(reopen_sla)
```

```python
reassignment_analysis = (
    incident_df
    .groupby("reassignment_count")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum"),
        median_resolution_hours=("resolution_time_hours", "median")
    )
    .reset_index()
)

reassignment_analysis["sla_breach_rate"] = (
    reassignment_analysis["sla_breaches"] /
    reassignment_analysis["total_incidents"] * 100
)

display(reassignment_analysis.head(15))
```

```python
state_analysis = (
    incident_df
    .groupby("final_state")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum"),
        median_resolution_hours=("resolution_time_hours", "median")
    )
    .reset_index()
)

state_analysis["sla_breach_rate"] = (
    state_analysis["sla_breaches"] /
    state_analysis["total_incidents"] * 100
)

display(
    state_analysis.sort_values(
        "total_incidents",
        ascending=False
    )
)
```

```python
incident_df["opened_month"] = (
    incident_df["opened_at"]
    .dt.to_period("M")
    .astype(str)
)

monthly_volume = (
    incident_df
    .groupby("opened_month")
    .agg(
        total_incidents=("number", "count"),
        sla_breaches=("sla_breached", "sum"),
        median_resolution_hours=("resolution_time_hours", "median")
    )
    .reset_index()
)

monthly_volume["sla_breach_rate"] = (
    monthly_volume["sla_breaches"] /
    monthly_volume["total_incidents"] * 100
)

display(monthly_volume)
```
